# GenIce3 → CS2 clathrate (Yaplot)

**Runtime → Run all** だけで、structure II クラスレート（**CS2**）にメタンを入れた構造を作り、[Yaplot](https://vitroid.github.io/Yaplot/) で表示します。

```python
from genice3.genice import GenIce3
from genice3.plugin import Exporter
from genice3.cli.options import parse_guest_option
from yaplot_view import show

# CS2 = sII clathrate; A16 = large cages → methane (me)
genice = GenIce3(
    seed=1,
    guests=parse_guest_option({"A16": "me"}),
)
genice.set_unitcell("CS2")

# H=1 → draw water molecules; click the canvas for Yaplot keys
show(Exporter("yaplot").dumps(genice, H=1), height=520)
```
