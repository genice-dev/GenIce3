# GenIce3 → CS2 clathrate (Yaplot)

**Runtime → Run all** だけで、structure II クラスレート（**CS2**）にメタンを入れた構造を作り、[Yaplot](https://vitroid.github.io/Yaplot/) で表示します。


# Installation

On Google Colaboratory, install GenIce3 (includes `yaplotlib`). Everywhere, fetch the Yaplot notebook helper if missing.


```python
from pathlib import Path
import urllib.request

try:
    import google.colab  # noqa: F401
    %pip install -q genice3
except ImportError:
    pass  # local Jupyter: assume genice3 is already installed

# yaplotlib comes with genice3; only the WASM notebook helper is extra
helper = Path("yaplot_view.py")
if not helper.is_file():
    urllib.request.urlretrieve(
        "https://vitroid.github.io/Yaplot/yaplot_view.py", helper
    )
print("yaplot_view ready:", helper.resolve())
```

```python
from genice3.genice import GenIce3
from genice3.plugin import Exporter
from genice3.cli.options import parse_guest_option
from yaplot_view import show

# CS2 = sII clathrate; A16 = large cages → methane (me)
genice = GenIce3(
    seed=1,
    guests=parse_guest_option({"A16": "me"}),
)
genice.set_unitcell("CS2")

# H=1 → draw water molecules; click the canvas for Yaplot keys
show(Exporter("yaplot").dumps(genice, H=1), height=520)
```
